# Kahaani evaluation results

Rebuilds `eval/out/report.md` and the charts from whatever result files exist, then shows them.
Run from the repo root kernel, with the eval group: `uv run --group eval jupyter lab` (or VS Code).

```python
import os
import sys
from pathlib import Path

ROOT = Path.cwd() if (Path.cwd() / "eval").exists() else Path.cwd().parent
os.chdir(ROOT)
sys.path.insert(0, str(ROOT))

from eval.report import build_report  # noqa: E402

text = build_report()
```

```python
from IPython.display import Markdown, display

display(Markdown(text.split("## Charts")[0]))
```

```python
from IPython.display import Image as Img

for png in sorted(Path("eval/out/charts").glob("*.png")):
    print(png.name)
    display(Img(filename=str(png)))
```

```python
import pandas as pd

clip = Path("eval/out/consistency_clip.jsonl")
if clip.exists():
    df = pd.read_json(clip, lines=True)
    display(df.pivot_table(index="id", columns="setup", values="ref_sim_mean").round(3))
```
